### Câu 1 (2 điểm): Cho một ảnh bất kỳ (tên ảnh do sinh viên tự đặt, ví dụ: `my_image.jpg`) và thực hiện các yêu cầu sau:

* Viết chương trình sử dụng Bilateral filter để làm mịn ảnh. (0.5 điểm)  
* Viết chương trình sử dụng Canny Edge Detection để xác định biên của hình ảnh. (0.5 điểm)  
* Đổi màu ảnh bằng cách hoán đổi kênh màu theo thứ tự (ví dụ: BGR → BRG) và lưu thành tên dạng `[ten_anh]_swapped.jpg`. (0.5 điểm)  
* Chuyển ảnh sang không gian màu YCrCb và tách riêng 3 kênh Y, Cr, Cb, lưu thành ảnh grayscale tương ứng (`[ten_anh]_Y.jpg`, `[ten_anh]_Cr.jpg`, `[ten_anh]_Cb.jpg`). (0.5 điểm)


```python
import cv2
import os

def process_to_ycrcb_and_save(image_path):
    """
    Đọc ảnh, chuyển sang YCrCb, tách Y, Cr, Cb và lưu từng kênh grayscale.
    Tên file đầu ra: [base_name]_Y.jpg, [base_name]_Cr.jpg, [base_name]_Cb.jpg
    """
    img = cv2.imread(image_path)
    if img is None:
        print(f"❌ Không đọc được ảnh: {image_path}")
        return

    base_name = os.path.splitext(os.path.basename(image_path))[0]
    ycrcb = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)
    Y, Cr, Cb = cv2.split(ycrcb)

    # Lưu từng kênh
    cv2.imwrite(f"{base_name}_Y.jpg", Y)
    cv2.imwrite(f"{base_name}_Cr.jpg", Cr)
    cv2.imwrite(f"{base_name}_Cb.jpg", Cb)

    print(f"✅ Đã lưu: {base_name}_Y.jpg, {base_name}_Cr.jpg, {base_name}_Cb.jpg")


def process_folder(folder_path):
    """
    Xử lý tất cả ảnh trong thư mục theo định dạng jpg/jpeg/png.
    """
    if not os.path.isdir(folder_path):
        print(f"❌ Đường dẫn không hợp lệ: {folder_path}")
        return

    for fname in os.listdir(folder_path):
        if fname.lower().endswith(('.jpg', '.jpeg', '.png')):
            process_to_ycrcb_and_save(os.path.join(folder_path, fname))


if __name__ == "__main__":
    #❶ Cách 1: xử lý thư mục (tất cả ảnh trong folder)
    folder = input("Nhập đường dẫn thư mục chứa ảnh: ").strip()
    if folder:
        process_folder(folder)
    else:
        #❷ Cách 2: xử lý danh sách ảnh cụ thể
        filenames = [
            "ten_anh_1.jpg",
            "ten_anh_2.jpg",
            "ten_anh_3.jpg"
        ]
        for fn in filenames:
            process_to_ycrcb_and_save(fn)
```

### Câu 2 (4 điểm) Viết một chương trình Python sử dụng OpenCV để tạo menu tương tác cho phép người dùng chọn các kỹ thuật biến đổi hình học và xử lý ảnh nâng cao từ một danh sách, áp dụng đồng thời cho nhiều ảnh.

### Yêu cầu:

1. Menu gồm:  
* Phóng to ảnh (Zoom bằng resize) (0.5 điểm)  
* Xoay ảnh (góc ngẫu nhiên từ 0–360 độ) (0.5 điểm)  
* Lật ảnh ngang (0.5 điểm)  
* Lật ảnh dọc (0.5 điểm)  
* Cắt ảnh (crop ngẫu nhiên vùng giữa ảnh) (0.5 điểm)  
* Thêm viền (padding màu ngẫu nhiên) (0.5 điểm)

2. Chương trình xử lý đồng thời 3 ảnh bất kỳ do sinh viên tự chọn (có thể chọn bằng đường dẫn file hoặc nhập tên ảnh tùy ý). (0.5 điểm)

3. Phím tương ứng để kích hoạt các phương pháp xử lý:  
* Z: Zoom  
* T: Rotate  
* H: Horizontal Flip  
* V: Vertical Flip  
* C: Crop  
* P: Padding (0.5 điểm)

4. Lưu file kết quả với định dạng: `result_[phương pháp]_[tên ảnh gốc].jpg`  
   Ví dụ: `result_crop_cat.jpg`, `result_rotate_image1.jpg` (0.5 điểm)


```python
import cv2
import os
import random
import numpy as np

# ================== Các hàm xử lý ==================

def zoom_image(image):
    scale = random.uniform(1.2, 2.0)
    height, width = image.shape[:2]
    resized = cv2.resize(image, (int(width * scale), int(height * scale)))
    return resized

def rotate_image(image):
    angle = random.uniform(0, 360)
    h, w = image.shape[:2]
    center = (w // 2, h // 2)
    matrix = cv2.getRotationMatrix2D(center, angle, 1.0)
    rotated = cv2.warpAffine(image, matrix, (w, h), borderMode=cv2.BORDER_REFLECT)
    return rotated

def horizontal_flip(image):
    return cv2.flip(image, 1)

def vertical_flip(image):
    return cv2.flip(image, 0)

def crop_image(image):
    h, w = image.shape[:2]
    top = random.randint(0, h // 4)
    bottom = random.randint(3 * h // 4, h)
    left = random.randint(0, w // 4)
    right = random.randint(3 * w // 4, w)
    return image[top:bottom, left:right]

def add_padding(image):
    top = random.randint(10, 50)
    bottom = random.randint(10, 50)
    left = random.randint(10, 50)
    right = random.randint(10, 50)
    color = [random.randint(0, 255) for _ in range(3)]
    padded = cv2.copyMakeBorder(image, top, bottom, left, right, cv2.BORDER_CONSTANT, value=color)
    return padded

# ================== Map ký hiệu với hàm xử lý ==================

methods = {
    'Z': ("zoom", zoom_image),
    'T': ("rotate", rotate_image),
    'H': ("hflip", horizontal_flip),
    'V': ("vflip", vertical_flip),
    'C': ("crop", crop_image),
    'P': ("padding", add_padding)
}

# ================== Giao diện chính ==================

def main():
    path = input("Nhập đường dẫn đến ảnh hoặc thư mục ảnh: ").strip()

    # Xử lý ảnh đơn hoặc nhiều ảnh từ thư mục
    if os.path.isfile(path):
        images = [path]
    elif os.path.isdir(path):
        images = [os.path.join(path, f) for f in os.listdir(path)
                  if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    else:
        print("Đường dẫn không hợp lệ.")
        return

    print("Chọn 3 kỹ thuật xử lý bằng cách nhập 3 phím ví dụ: ZTV (Zoom, Rotate, Vertical Flip)")
    print("Phím: Z = Zoom | T = Rotate | H = Horizontal Flip | V = Vertical Flip | C = Crop | P = Padding")
    choices = input("Nhập 3 phím kỹ thuật: ").upper()

    if len(choices) != 3 or any(c not in methods for c in choices):
        print("Lựa chọn không hợp lệ!")
        return

    output_dir = "output_results"
    os.makedirs(output_dir, exist_ok=True)

    for img_path in images:
        img = cv2.imread(img_path)
        if img is None:
            print(f"Không thể đọc ảnh: {img_path}")
            continue

        base_name = os.path.splitext(os.path.basename(img_path))[0]

        for key in choices:
            method_name, func = methods[key]
            img = func(img)
            save_path = os.path.join(output_dir, f"result_{method_name}_{base_name}.jpg")
            cv2.imwrite(save_path, img)

    print("✅ Đã xử lý xong ảnh. Kết quả nằm trong thư mục 'output_results'.")

# ================== Chạy chương trình ==================

if __name__ == "__main__":
    main()
```

### Câu 3 (4 điểm) Viết một chương trình Python để xử lý 3 ảnh bất kỳ do sinh viên tự chọn.

* Thêm viền đen 20 pixel cho ảnh đầu tiên. (0.5 điểm)  
* Xoay ảnh thứ hai 45 độ và phóng to 1.5 lần. (0.5 điểm)  
* Tăng kích thước ảnh thứ ba lên 4 lần, sau đó áp dụng Bilateral Filter với tham số tùy chọn. (1.5 điểm)  
* Thay đổi độ sáng và độ tương phản ảnh thứ ba theo công thức:

$$
I_{out}(x, y) = \alpha \cdot I_{in}(x, y) + \beta
$$

Trong đó:  

$$
\alpha \in [0.6, 2.0], \quad \beta \in [-60, 60]
$$

Giá trị đầu ra cần được giới hạn trong khoảng [0, 255] bằng công thức:

$$
I_{out}(x, y) = \text{clip}(I_{out}(x, y), 0, 255)
$$

```python
import cv2
import numpy as np
import os

def read_image_unicode(path):
    with open(path, "rb") as f:
        data = f.read()
        img_array = np.asarray(bytearray(data), dtype=np.uint8)
        return cv2.imdecode(img_array, cv2.IMREAD_COLOR)

def add_black_border(img, border_size=20):
    return cv2.copyMakeBorder(img, border_size, border_size, border_size, border_size,
                              cv2.BORDER_CONSTANT, value=(0, 0, 0))

def rotate_and_scale(img, angle=45, scale=1.5):
    h, w = img.shape[:2]
    center = (w // 2, h // 2)
    M = cv2.getRotationMatrix2D(center, angle, scale)
    return cv2.warpAffine(img, M, (w, h))

def resize_and_bilateral_filter(img, resize_factor=4):
    resized = cv2.resize(img, None, fx=resize_factor, fy=resize_factor, interpolation=cv2.INTER_LINEAR)
    return cv2.bilateralFilter(resized, d=9, sigmaColor=75, sigmaSpace=75)

def adjust_brightness_contrast(img, alpha, beta):
    out = alpha * img + beta
    out = np.clip(out, 0, 255)
    return out.astype(np.uint8)

def main():
    paths = ["1.png", "2.png", "3.png"]  # Đảm bảo 3 ảnh này đã nằm trong thư mục đang chạy

    imgs = [read_image_unicode(p) for p in paths]
    output_dir = "output_cau3"
    os.makedirs(output_dir, exist_ok=True)

    # 1. Ảnh đầu tiên: thêm viền đen 20 pixel
    img1 = add_black_border(imgs[0])
    cv2.imwrite(os.path.join(output_dir, "image1_border.png"), img1)

    # 2. Ảnh thứ hai: xoay 45 độ và scale 1.5 lần
    img2 = rotate_and_scale(imgs[1])
    cv2.imwrite(os.path.join(output_dir, "image2_rotate_scale.png"), img2)

    # 3. Ảnh thứ ba: resize 4 lần rồi lọc Bilateral
    img3_resized_filtered = resize_and_bilateral_filter(imgs[2])
    cv2.imwrite(os.path.join(output_dir, "image3_resize_filter.png"), img3_resized_filtered)

    # 4. Thay đổi độ sáng và độ tương phản ảnh thứ ba (sau khi resize + filter)
    alpha = 1.8   # Thuộc khoảng [0.6, 2.0]
    beta = 40     # Thuộc khoảng [-60, 60]
    img3_bright = adjust_brightness_contrast(img3_resized_filtered, alpha, beta)
    cv2.imwrite(os.path.join(output_dir, "image3_brightness_contrast.png"), img3_bright)

    print("✅ Xử lý hoàn tất. Kết quả nằm trong thư mục 'output_cau3'.")

if __name__ == "__main__":
    main()
```
